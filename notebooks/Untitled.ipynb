```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path(r"C:\Users\Harshini J\Engineering\Projects\Nimisha Ma'am - Project\mamba-dti\data\raw\human_random")

train = pd.read_csv(DATA_DIR / "train" / "sequence.csv")
valid = pd.read_csv(DATA_DIR / "valid" / "sequence.csv")
test  = pd.read_csv(DATA_DIR / "test" / "sequence.csv")
```

```python
proteins = pd.concat([
    train["sequence"],
    valid["sequence"],
    test["sequence"]
], ignore_index=True)
```

```python
proteins = (
    proteins
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
len(proteins)
```

```python
protein_df = pd.DataFrame({
    "protein_id": range(len(proteins)),
    "sequence": proteins
})
```

```python
protein_df.to_csv(
    r"C:\Users\Harshini J\Engineering\Projects\Nimisha Ma'am - Project\mamba-dti\data\processed\protein_lookup.csv",
    index=False
)
```

```python
from pathlib import Path

output = Path(r"C:\Users\Harshini J\Engineering\Projects\Nimisha Ma'am - Project\mamba-dti\data\processed\proteins.fasta")

with open(output, "w") as f:
    for _, row in protein_df.iterrows():

        f.write(f">protein_{row.protein_id}\n")
        f.write(row.sequence + "\n")
```
